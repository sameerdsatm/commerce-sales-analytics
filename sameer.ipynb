```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate a quick sample E-Commerce Dataset
np.random.seed(42)
data = {
    'OrderID': range(1001, 1101),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Home & Kitchen'], 100),
    'Sales': np.random.uniform(50, 1500, 100).round(2),
    'Quantity': np.random.randint(1, 5, 100),
    'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100)
}
df = pd.DataFrame(data)

# 2. Basic Data Inspection
print("--- First 5 Rows of Data ---")
display(df.head())

print("\n--- Missing Values Check ---")
print(df.isnull().sum())

# 3. Data Analysis: Total Sales by Category
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
print("\n--- Total Sales by Category ---")
display(category_sales)

# 4. Data Visualization
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))
sns.barplot(x='Category', y='Sales', data=category_sales, palette='viridis')
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales ($)')
plt.show()

# Monthly Trend Visualization
plt.figure(figsize=(8, 5))
monthly_sales = df.groupby('Month')['Sales'].sum().reset_index()
sns.lineplot(x='Month', y='Sales', data=monthly_sales, marker='o', color='b')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.show()
```
